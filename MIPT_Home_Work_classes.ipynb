## MathComplex (3 балла)

В этом задании вам нужно реализовать класс `MathComplex` для комплексных чисел. Они должны быть идентичны привычным нам комплексным числам из математики, а не объектам класса `complex`. Например, число `0+1i` должно печататься как `i`, число `42+0i` как просто `42`.

Вспомнить базовые свойства и операции комплексных чисел можно [здесь](http://www.mathprofi.ru/kompleksnye_chisla_dlya_chainikov.html) и [здесь](http://mathportal.net/index.php/kompleksnye-chisla/dejstviya-s-kompleksnymi-chislami).

Последовательно выполните указанные пункты:

1. реализовать методы `__init__` и нахождение модуля `__abs__`. Атрибуты назовите `real` и `imag`
2. красивый вывод на печать (достаточно реализовать `__repr__`). Для обозначения мнимой части используйте символ `i`
3. перегрузить операторы:
    - сложения
    - умножения
    - вычитания
    - деления (`__truediv__`)
<!--     - возведения в степень (считать, что только натуральную) -->

```python
class MathComplex:
  def __init__(self, real=0, imag=0):
    self.real = real
    self.imag = imag
    if not isinstance(real, (int, float)):
      raise Exception('Действительная часть комплексного числа должна быть числом')
    if not isinstance(imag, (int, float)):
      raise Exception('Мнимая часть комплексного числа должна быть числом')

  def __abs__(self):
    return (self.real **2 + self.imag ** 2) ** 0.5

  def __repr__(self):
    if self.imag == 0:
      return f'{self.real}'
    if self.real == 0 and self.imag == 1:
      return 'i'
    if self.real == 0 and self.imag == -1:
      return '-i'
    if self.real == 0:
      return f'{self.imag}'
    if self.imag == 1:
      return f'{self.real}+i'
    if self.imag == -1:
      return f'{self.real}-i'
    if self.imag > 0:
      return f'{self.real}+{self.imag}i'
    else:
      return f'{self.real}{self.imag}i'

  def __add__(self, other):
    new_real = self.real + other.real
    new_imag = self.imag + other.imag
    return MathComplex(new_real, new_imag)

  def __sub__(self, other):
    new_real = self.real - other.real
    new_imag = self.imag - other.imag
    return MathComplex(new_real, new_imag)

  def __mul__(self, other):
    new_real = self.real * other.real - self.imag * other.imag
    new_imag = self.real * other.imag + self.imag * other.real
    return MathComplex(new_real, new_imag)

  def __rmul__(self, other):
    new_real = self.real * other.real - self.imag * other.imag
    new_imag = self.real * other.imag + self.imag * other.real
    return MathComplex(new_real, new_imag)

  def __truediv__(self, other):
    new_real = (self.real * other.real + self.imag * other.imag) / (
               other.real ** 2 + other.imag ** 2)
    new_imag = (self.imag * other.real - self.real * other.imag) / (
               other.real ** 2 + other.imag ** 2)
    return MathComplex(new_real, new_imag)

  def __rtruediv__(self, other):
    new_real = (self.real * other.real + self.imag * other.imag) / (
               other.real ** 2 + other.imag ** 2)
    new_imag = (self.imag * other.real - self.real * other.imag) / (
               other.real ** 2 + other.imag ** 2)
    return MathComplex(new_real, new_imag)
```

## ExtendedComplex (1+1.5 балла)

4. Вам нужно отнаследовать класс `ExtendedComplex` и реализовать в нем соответствующие специальные методы, чтобы арифметические операции из пункта 2 выполнялись еще и с операндами типа `int/float`. Немного подробнее: если `а` - объект любого из классов `(int, float, ExtendedComplex`), `x` - экземпляр `ExtendedComplex`, то должны выполняться оба выражения `x + a` и `a + x`.

5. Тесты для этого задания надо написать самостоятельно в соответствующей ячейке. Также реализуйте в этом классе сравнение (одно из чисел может быть `int` или `float`) по следующим правилам:

    - равны, если одновременно равны действительные и мнимые части
    - одно число больше другого, если действительная или мнимая часть первого больше соответствующей части второго, а оставшаяся часть не меньше
    - во всех оставшихся случаях выводите сообщение: `"результат не определен"`

**Подсказки**: `isinstance`, специальные методы для сравнения и правых выражений `a + x`, например, `__radd__`, для 5 рекомендуется начать с написания тестов на каждый оператор сравнения.

```python
class ExtendedComplex(MathComplex):
    def __mul__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(num * self.real, num * self.imag)
      else:
        return super().__mul__(self)

    def __rmul__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(num * self.real, num * self.imag)
      else:
        return super().__rmul__(self)

    def __add__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(num + self.real, self.imag)
      else:
        return super().__add__(self)

    def __radd__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(num + self.real, self.imag)
      else:
        return super().__add__(self)

    def __sub__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(self.real - num, self.imag)
      else:
        return super().__sub__(self)

    def __rsub__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(num - self.real, -self.imag)
      else:
        return super().__sub__(self)

    def __truediv__(self, num):
      if isinstance(num, (float, int)):
        return ExtendedComplex(self.real / num, self.imag / num)
      else:
        return super().__truediv__(self)

    def __rtruediv__(self, num):
      if isinstance(num, (float, int)):
        conj = ExtendedComplex(self.real, self.imag)
        numerator = conj.__mul__(num)
        self = numerator
        denominator = super().__mul__(conj)
        return num * super().__truediv__(denominator)
      else:
        return super().__rtruediv__(self)

    def __eq__(self, other):
      return  self.real == other.real and self.imag == other.imag

    def __gt__(self, other):
      if (self.real == other.real and self.imag >= other.imag) or (
          self.imag == other.imag and self.real >= other.real):
        return True
      else:
        return 'результат неопределен'
```

## MyRange (5 баллов)

6. Реализуйте класс `Range` и тесты к нему, который будет аналогом обычного `range`. Подробности в ячейке ниже: ориентируйтесь на перечисленные магические методы, которые нужно реализовать. Вам понадобится также сделать `RangeIterator`, экземпляры которого должны возвращаться методами `__iter__` и `__reversed__`. Проверка вхождения не должна вызывать итератор!

**Указание:** реализация возможности брать срез **целиком аналогично** обычному **range** требует дополнительных пересчетов индексов! В примерах ниже такие случаи специально опущены - поищите их самостоятельно :) Вспомните, что возвращает взятие среза в различных случаях.

```python
class RangeIterator:
    def __init__(self, rangeobj):
      self.start = rangeobj.start
      self.stop = rangeobj.stop
      self.step = rangeobj.step

    def __next__(self):
      if (self.step > 0 and self.start >= self.stop) or (
          self.step < 0 and self.start <= self.stop):
          raise StopIteration
      result = self.start
      self.start += self.step
      return result

    def __iter__(self):
      return self


class Range:
    def __init__(self, *args):
      if len(args)==1:
          self.start = 0
          self.stop = args[0]
          self.step = 1
      elif len(args) == 2:
          self.start, self.stop = args
          self.step = 1
      elif len(args) == 3:
          self.start, self.stop, self.step = args
      else:
          raise TypeError(f"Range expected 1-3 arguments, got {len(args)}")
      if self.step == 0:
          raise ValueError("Range() arg 3 must not be zero")

    def __eq__(self, other):
      return (self.start == other.start and
              self.stop == other.stop and
              self.step == other.step)

    def __iter__(self):
      return RangeIterator(self)

    def __reversed__(self):
      n = len(self)
      if n == 0:
          return iter(())
      new_start = self.start + (n - 1) * self.step
      new_step = -self.step
      new_stop = self.start - self.step
      return iter(Range(new_start, new_stop, new_step))

    def __repr__(self):
      if self.start and self.stop and self.step !=1:
        return f'{self.__class__.__name__}({self.start}, {self.stop}, {self.step})'
      if self.start and self.stop:
        return f'{self.__class__.__name__}({self.start}, {self.stop})'
      if self.stop:
        return f'{self.__class__.__name__}({self.stop})'

    def __getitem__(self, index):
      if isinstance(index, slice):
        length = len(self)
        start, stop, step = index.indices(length)
        indices = Range(start, stop, step)
        n = len(indices)
        if n == 0:
          return Range(0, 0, 1)

        first = self.start + start * self.step
        step_new = self.step * step
        last_index = start + (n - 1) * step
        last = self.start + last_index * self.step
        new_stop = last + step_new
        return Range(first, new_stop, step_new)

      elif isinstance(index, int):
          if index < 0:
              index += len(self)
          if index < 0 or index >= len(self):
              raise IndexError("Range object index out of range")
          return self.start + index * self.step
      else:
          raise TypeError("Range indices must be integers or slices")

    def __len__(self):
      if (self.step > 0 and self.start < self.stop) or (
          self.step < 0 and self.start > self.stop):
          difference = self.stop - self.start
          if self.step > 0:
              return (difference - 1) // self.step + 1
          else:
              return (difference + 1) // self.step + 1
      else:
          return 0

    def __contains__(self, value):
      if not isinstance(value, int):
        return False
      if value >= self.start and value < self.stop:
        return True
```
