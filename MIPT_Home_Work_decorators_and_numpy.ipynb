- параметры `ans` и `gt` - строго позиционные
- параметр `ndigits` - необязательный, строго именованный. Если он задан, то значение, возвращаемое функцией, округляется до `ndigits` знаков после запятой.

**Примечание:**

- ans - сокращение от answer - проставленные ответы, список из 0 и 1
- gt - сокращение от ground truth - правильные ответы, список из 0 и 1
- каждая задача этой части по 0.3 балла

## Задача 1

Напишите функцию `accuracy(ans, gt, ndigits)`, считающую долю правильных ответов.

````python
def accuracy(ans, gt, /, *, ndigits=None):
  right = 0
  for i, j in zip(ans, gt):
    if i == j:
      right += 1
    else:
      continue
  if ndigits is None:
    return right / len(gt)
  else:
    return round(right / len(gt), ndigits)
````

## Задача 2

Напишите функцию `true_positive_rate(ans, gt, ndigits)`, считающую долю правильно отмеченных единиц.

````python
def true_positive_rate(ans, gt, /, *, ndigits=None):
  right = 0
  for i, j in zip(ans, gt):
    if i == j and i ==1:
      right += 1
    else:
      continue
  if ndigits is None:
    return right / len(gt)
  else:
    return round(right / len(gt), ndigits)
````

## Задача 3

Напишите функцию `false_positive_rate(ans, gt, ndigits)`, считающую долю неправильно отмеченных единиц.

````python
def false_positive_rate(ans, gt, /, *, ndigits=None):
  right = 0
  for i, j in zip(ans, gt):
    if i != j and i ==1:
      right += 1
    else:
      continue
  if ndigits is None:
    return right / len(gt)
  else:
    return round(right / len(gt), ndigits)
````

## Задача 6

Напишите декоратор `timing`, который будет считать время выполнения последнего вызова функции в секундах и сохранять его в атрибуте `last_time`. Для работы со временем используйте встроенный модуль.

````python
from time import time

def timing(func):
  start = time()
  def wrapper(*args, **kwargs):
    wrapper.last_time = time() - start
    return func(*args, **kwargs)
  return wrapper
````

## Задача 7

Напишите декоратор `calls_counter`, который будет считать число рекурсивных вызовов функции во время ее последнего вызова и сохранять в атрибуте `calls`.

````python
def calls_counter(func):
  def wrapper(*args, **kwargs):
    if not hasattr(wrapper, 'calls'):
      wrapper.calls = 0
    wrapper.calls +=1
    return func(*args, **kwargs)
  return wrapper
````

## Задача 8

Придумайте функцию, подходящую для тестирования предыдущих декораторов и протестируйте их.

````python
@timing
def factorial(n):
  if n <= 1:
    return 1
  else:
    return n * factorial(n-1)

factorial(10)
f_10 = factorial.last_time
factorial(1000)
f_1000 = factorial.last_time

assert f_10 < f_1000
````

````python
@calls_counter
def factorial(n):
  if n <= 1:
    return 1
  else:
    return n * factorial(n-1)

factorial(5)
assert factorial.calls == 5
````

Подсказка: `__new__`

````python
def single(deprecated_cls):
  class Singleton:
      _instance = None

      def __new__(cls):
          if cls._instance is None:
              cls._instance = super().__new__(cls)
          return cls._instance
  return Singleton
````

````python
from collections import deque

data_lst = list()
search_queue = deque()
search_queue.append(extract_to_path)
while search_queue:
    dir = search_queue.popleft()
    for file in sorted(os.listdir(dir), key = lambda x: int(x[3:]) if x.startswith('dir') else -1):
        full_path = os.path.join(dir, file)
        if os.path.isfile(full_path):
            try:
                with open(full_path, 'r', encoding='utf-8') as rd:
                    elem = rd.read()
                    data_lst.append(elem)
            except:
                continue  
        else:
            search_queue.append(full_path)
````
