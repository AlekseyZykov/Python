### Задача 6 (0.5). Расширение файла

Вы получаете на вход имя файла. Необходимо вывести его расширение. Если расширения у файла нет, выводите `"Нет расширения"`.

Под разрешением файла понимают определённую последовательность букв или цифр, отделенную от основного имени файла точкой.

**Примечание:**
- для файла `not_a_virus.jpg.exe` расширением является `.exe`, а не `.jpg.exe`.

#### Пример

**Ввод:**

```
absolutely_not_a_virus.exe
```

**Вывод:**

```
.exe
```

````python
filename = input("Введите имя файла: ")
````

````python
for i in range(len(filename)):
  if filename[i] == '.':
    extension = ''
    for j in range(i, len(filename)):
      extension += filename[j]
print(extension)
````

````python
counter_message = 0
agg_price = 0
while True:
  counter_message += 1
  message = input()
# Условие выхода из цикла
  if message =='#':
    break
# Обработка первого сообщения
  if counter_message == 1:
    price = float(message)
    min_price = price
    max_price = price
# Обработка сообщений содержащих МЕНЬШЕ
  if 'МЕНЬШЕ' in message:
# Проверка отсутствия ошибок
    if price - float(message[6:]) > 0:
      price -= float(message[6:])
# Запись минимальной цены
      if min_price > price:
        min_price = price
# Обработка ошибок
    else:
      continue
# Обработка сообщений содержащих БЕЗ ИЗМЕНЕНИЙ
  if 'БЕЗ ИЗМЕНЕНИЙ' in message:
    continue
# Обработка сообщений содержащих БОЛЬШЕ
  if 'БОЛЬШЕ' in message:
    price += float(message[6:])
# Запись максимальной цены
    if max_price < price:
      max_price = price
# Агрегируем цены и находим среднее
  agg_price += price
  average_price = agg_price / counter_message
# Принимаем текущую цену от брокера Александра Ивановича
  current_price = 0.6 * min_price + 0.4 * max_price
# Проверяем условия и отправляем брокеру сообщение
  if 2 * current_price < average_price:
    print('КУПИТЬ_10', file = sys.stderr)
  elif 1.5 * current_price < average_price:
    print('КУПИТЬ_7', file = sys.stderr)
  elif current_price > 1.5 * average_price:
    print('ПРОДАТЬ_6', file = sys.stderr)
  elif current_price > 1.2 * average_price:
    print('ПРОДАТЬ_3', file = sys.stderr)
  else:
    print('ДЕРЖАТЬ', file = sys.stderr)
````

### Задача 10 (1). Частичный разворот

Вы принимаете на вход строку. Необходимо:

1. создать ее копию, в которой символы исходной с **нечетными индексами** переставлены в обратном порядке
2. вывести получившуюся строку

#### Пример

**Ввод:**

```
0123456789
```

**Вывод:**

```
0927456381
```

````python
string = input("Введите строку: ")
````

````python
new_string = ''
reversed_odd = string[-1::-2]
even = string[0::2]
cnt_odd = 0
cnt_even = 0
for i in range(len(string)):
  if i % 2 ==0:
    new_string += even[cnt_even]
    cnt_even += 1
  else:
    new_string += reversed_odd[cnt_odd]
    cnt_odd += 1
print(new_string)
````
