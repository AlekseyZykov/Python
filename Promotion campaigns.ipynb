### **Самая эффективная платформа по средним значениям конверсий**

Результат сохраняется в переменную `most_effective_platform`

````python
import csv

# Читаем csv файл
with open('campaign_data.csv', 'r') as csvfile:
    csvreader = csv.reader(csvfile)
    for i, row in enumerate(csvreader):
        # Записываем названия столбцов в отдельный список
        if i == 0:
            headers = row
# Получаем данные в виде списка словарей
        data_dict = [dict(zip(headers, row)) for row in csvreader]
# Получаем множество платформ
    platforms = set()
    for elem in data_dict:
        platforms.add(elem['Платформа'])
# Создаем словарь платформ и конверсий
    dict_conv = {}
    for platform in platforms:
        # Разбиваем (фильтруем) конверсии по платформам
        res = list(filter(lambda x: platform == x['Платформа'], data_dict))
        conv = 0
# Инициализируем счетчик рекламных кампаний по платформам
        i = 1
        for elem in res:
            # Агрегируем конверсии
            conv += int(elem.get('Конверсия'))
# Находим среднюю конверсию
            dict_conv[platform] = conv/i
            i += 1
most_effective_platform = max(dict_conv, key=dict_conv.get)
````

### **Доля затрат на рекламные кампании каждой платформы и каждого города**

Результат сохраняется в файл `platform_city_results.txt`

Результат имеет такой же порядок групп, что и исходный файл. Города внутри платформ отсортированы по убыванию доли затрат. Доля затрат округлена до 2 знаков после запятой.

Результат выглядит примерно так:

```
Для группы Telegram:
- Город: Москва, доля затрат на рекламу: 16.46%
- Город: Калуга, доля затрат на рекламу: 16.2%
- Город: Казань, доля затрат на рекламу: 14.57%
- ...
Для группы Вконтакте:
- Город: Екатеринбург, доля затрат на рекламу: 17.43%
- Город: Новосибирск, доля затрат на рекламу: 14.86%
- Город: Санкт-Петербург, доля затрат на рекламу: 14.13%
...
```

````python
import csv

# Читаем csv файл
with open('campaign_data.csv', 'r') as csvfile:
    csvreader = csv.reader(csvfile)
    for i, row in enumerate(csvreader):
        # Записываем названия столбцов в отдельный список
        if i == 0:
            headers = row
# Получаем данные в виде списка словарей
        data_dict = [dict(zip(headers, row)) for row in csvreader]
# Получаем список платформ и городов
platforms = []
for elem in data_dict:
    if elem not in platforms:
        platforms.append(elem['Платформа'])
cities = []
for elem in data_dict:
    if elem not in cities:
        cities.append(elem['Город'])
# Создаем словарь платформ
platform_cities_expenses = {}
for platform in platforms:
    # Создаем словарь городов для конкретной платформы
    city_expenses = {}
# Разбиваем (фильтруем) расходы по платформам
    # сформирован список по конкретной платформе
    plat = list(filter(lambda x: platform == x['Платформа'], data_dict))
# Разбиваем (фильтруем) расходы по городам
    for city in cities:
        # сформирован список городов на конкретной платформе
        ct = list(filter(lambda x: city == x['Город'], plat))
        expenses = 0
        for elem in ct:
            # Агрегируем расходы по городам и записываем в словарь
            # считаем расходы по конкретному городу
            expenses += int(elem.get('Бюджет'))
        city_expenses[city] = expenses
# Записываем в словарь платформ расходы по городам
    platform_cities_expenses[platform] = city_expenses
# Находим затраты по платформам без городов
platform_expenses = {
    k: sum(platform_cities_expenses[k].values()) for k in platform_cities_expenses}
# Записываем в файл
with open('platform_city_results.txt', 'w') as result:
    for platform in platform_expenses:
        # Для каждой платформы сортируем словарь городов и расходов
        platform_cities_expenses_sorted = dict(sorted(
            platform_cities_expenses[platform].items(), key=lambda item: item[1], reverse=True))
# Получаем список городов
        cities = list(platform_cities_expenses_sorted.keys())
        result.write(f'Для группы {platform}:\n')
# Построчно записываем результат
        for city in cities:
            expense = round(
                100 * platform_cities_expenses_sorted[city] / platform_expenses[platform], 2)
            result.write(
                f'- Город: {city}, доля затрат на рекламу: {expense}%\n')
````

### **Функция, которая возвращает средний бюджет всех кампаний, запущенных на определенной платформе**

Функция `calculate_average_budget` принимает путь к файлу и название плафтормы и возвращает число - средний бюджет всех кампаний, запущенных на данной платформе, округленный до двух знаков после запятой.

````python
import csv


def calculate_average_budget(path, platform):
    # Читаем csv файл
    with open(path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for i, row in enumerate(csvreader):
            # Записываем названия столбцов в отдельный список
            if i == 0:
                headers = row
# Получаем данные в виде списка словарей
            data_dict = [dict(zip(headers, row)) for row in csvreader]
# Разбиваем (фильтруем) бюджет по платформам
    budget_platfrom = list(
        filter(lambda x: platform == x['Платформа'], data_dict))
    budget = 0
    i = 0
    for elem in budget_platfrom:
        # Агрегируем бюджет
        budget += int(elem.get('Бюджет'))
        i += 1
# Находим средний бюджет по платформе
    res = round(budget / i, 2)
    return res
````

### **Генератор, который генерирует даты, в которые не запускались кампании**

Функция `get_missing_campaign_dates` принимает путь к файлу и генерирует построчно даты, отсортированные по возрастанию, в которые не запускались кампании, в формате:

```
2022-02-29
2022-03-28
2022-03-29
2022-05-30
2022-06-13
2022-06-27
...
```

````python
import csv
from datetime import datetime, timedelta


def get_missing_campaign_dates(path):
    # Читаем csv файл
    with open(path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for i, row in enumerate(csvreader):
            # Записываем названия столбцов в отдельный список
            if i == 0:
                headers = row
# Получаем данные в виде списка словарей
            data_dict = [dict(zip(headers, row)) for row in csvreader]
# Получаем множество дат запуска рекламных кампаний
        start_dates = set([datetime.strptime(
            elem['Начальная дата'], '%Y-%m-%d').date() for elem in data_dict])
# Генерируем календарь
        calendar = []
        end_date = min(start_dates)
        while end_date < max(start_dates):
            end_date += timedelta(days=1)
            calendar.append(end_date)
# Убираем дубли
        calendar = set(calendar)
# Находим дни, в которое не запускались рекламные кампании
        days_without_campaign = list(calendar - start_dates)
# Сортируем список
    days_without_campaign.sort()
    for i in days_without_campaign:
        yield i
````

### **Функция `group_campaign_data`принимает на вход путь к файлу и возвращает сгруппированные данные по городам, содержащие суммарный бюджет и количество кликов**

Функция `group_campaign_data` возвращает список словарей, отсортированный по городам, в формате:

```python
[{'Город': 'Нижний Новгород',
  'Количество кликов': 616.0,
  'Суммарный бюджет': 27585},
 {'Город': 'Новосибирск',
  'Количество кликов': 733.0,
  'Суммарный бюджет': 33001},
  ...]
```

````python
import csv


def group_campaign_data(path):
    # Читаем csv файл
    with open(path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for i, row in enumerate(csvreader):
            # Записываем названия столбцов в отдельный список
            if i == 0:
                headers = row
    # Получаем данные в виде списка словарей
            data_dict = [dict(zip(headers, row)) for row in csvreader]
# Получаем множество городов
    cities = set()
    for elem in data_dict:
        cities.add(elem['Город'])
    cities = sorted(list(cities))
# Разбиваем метрики по городу
    group_data = []
    for city in cities:
        data_filtered = list(filter(lambda x: city == x['Город'], data_dict))
        clicks = 0
        budget = 0
# Считаем метрики и выводим результат
        for elem in data_filtered:
            clicks += float(elem.get('Клики'))
            budget += int(elem.get('Бюджет'))
        group_data.append(
            {'Город': city, 'Количество кликов': clicks, 'Суммарный бюджет': budget})
    return group_data
````

### **Генератор, которы генерирует данные о кампаниях, запущенных в указанном городе и имеющие бюджет выше заданного значения**

Функция `campaign_generator` принимает путь к файлу, город и бюджет и генерирует словари, отсортированные по возрастанию 'ID Кампании' в формате:

```
{'ID Кампании': '657', 'Тип': 'Контекстная', 'Платформа': 'Вконтакте', 'Доход': '10692'}
{'ID Кампании': '674', 'Тип': 'Нативная интеграция', 'Платформа': 'Яндекс', 'Доход': '20766'}
{'ID Кампании': '761', 'Тип': 'Соцсети', 'Платформа': 'Вконтакте', 'Доход': '15680'}
```

````python
import csv


def campaign_generator(path, city, budget):
    # Читаем csv файл
    with open(path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for i, row in enumerate(csvreader):
            # Записываем названия столбцов в отдельный список
            if i == 0:
                headers = row
    # Получаем данные в виде списка словарей
            data_dict = [dict(zip(headers, row)) for row in csvreader]
# Фильтруем по указанному городу и бюджету
    data_filtered = list(
        filter(lambda x: city == x['Город'] and budget < int(x['Бюджет']), data_dict))
# Получаем нужный формат данных
    for row in data_filtered:
        yield {'ID Кампании': row['ID Кампании'], 'Тип': row['Тип кампании'], 'Платформа': row['Платформа'], 'Доход': row['Доход']}
````
